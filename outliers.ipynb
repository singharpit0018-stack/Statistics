```python
import numpy as np
```

```python
arr=np.array([1,2,2,7,8,9,10,12,12,13,14,14,15,20,21,21,25,25,30,30,120])
arr
```

```python
Q1=np.percentile(arr,25)
Q3=np.percentile(arr,75)
```

```python
IQR=Q3-Q1
IQR
```

```python
UF=Q3+(1.5*IQR)
LF=Q1-(1.5*IQR)
```

```python
l=[]
for i in arr:
    if i>=LF and i<=UF:
        l.append(i)
        arr2=np.array(l)
```
